## 問3: K-means（10点）

K-means法は，クラスタの初期化方法によってクラスタリングの結果が変わる．
したがって，適切な初期値を設定することが重要である．

クラスタの初期化方法として，以下の手順を考えてみよう：
1. クラスタ数に応じて，初期のクラスタ中心を適当に定める．
2. 各データについて，最も近い初期クラスタ中心が属するクラスタを初期クラスタとして割り当てる．

次に，以下の1次元データ集合をK-means法を用いて3つのクラスタに分割することを考える：

`1, 2, 3, 4, 5, 8, 9, 10, 11, 12, 24, 28, 32, 36, 40`

このデータ集合に対してK-means法を適用する際，（3つある）クラスタ中心の初期値として`{1, 2, 3}`を選んだ場合と`{1, 11, 28}`を選んだ場合のクラスタリング結果を比較しなさい．
また，どのようなデータをクラスタ中心の初期値に設定すれば， 直感的に妥当なクラスタリング結果が得られるかを考察しなさい．
